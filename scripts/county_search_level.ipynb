```python
import pandas as pd


df = pd.read_csv('NRI_Table_Counties.csv')


df.columns = df.columns.str.strip().str.upper()


if 'REPL_SCORE' in df.columns and 'RESL_SCORE' not in df.columns:
    df.rename(columns={
        'REPL_SCORE': 'RESL_SCORE', 
        'REPL_RATNG': 'RESL_RATNG'
    }, inplace=True)


output_filename = 'NRI_Table_Counties_FINAL.csv'
df.to_csv(output_filename, index=False)

print(f"Data cleaning successfully completed! Cleaned dataset saved as: {output_filename}")
```

```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output

def run_fema_risk_dashboard():
    # 1. 读取清洗后的 FINAL 数据集
    df = pd.read_csv('NRI_Table_Counties_FINAL.csv')
    df.columns = [str(c).strip() for c in df.columns]
    
    print("=" * 80)
    print("--- 1. TOP 10 COUNTIES BY TOTAL LOSS VALUE (EAL_VALT) ---")
    print("=" * 80)
    top_loss_cols = ['COUNTY', 'STATEABBRV', 'EAL_VALT', 'EAL_SCORE', 'EAL_RATNG', 'RESL_SCORE', 'RESL_RATNG']
    top_loss_df = df.nlargest(10, 'EAL_VALT')[top_loss_cols].reset_index(drop=True)
    display(top_loss_df)
    print("\n")
    
    print("=" * 80)
    print("--- 2. TOP 10 COUNTIES BY RESILIENCE SCORE (RESL_SCORE) ---")
    print("=" * 80)
    top_resl_cols = ['COUNTY', 'STATEABBRV', 'RESL_SCORE', 'RESL_RATNG', 'EAL_VALT', 'EAL_SCORE', 'EAL_RATNG']
    top_resl_df = df.nlargest(10, 'RESL_SCORE')[top_resl_cols].reset_index(drop=True)
    display(top_resl_df)
    print("\n")
    
    print("=" * 80)
    print("--- 3. INTERACTIVE COUNTY RISK & RESILIENCE SEARCH TOOL ---")
    print("=" * 80)
    
    search_cols = ['COUNTY', 'STATEABBRV', 'EAL_VALT', 'EAL_SCORE', 'EAL_RATNG', 'RESL_SCORE', 'RESL_RATNG']
    
    county_input = widgets.Text(
        placeholder='Type county name (e.g., Los Angeles, Cook)...',
        description='County Name:',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='450px')
    )
    
    state_input = widgets.Text(
        placeholder='Type state abbreviation (e.g., CA, IL)...',
        description='State Abbr:',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='450px')
    )
    
    search_output = widgets.Output()

    def update_search_results(*args):
        with search_output:
            clear_output()
            q_county = county_input.value.strip().lower()
            q_state = state_input.value.strip().lower()
            
            matches = df.copy()
            if q_county:
                matches = matches[matches['COUNTY'].astype(str).str.lower().str.contains(q_county)]
            if q_state:
                matches = matches[matches['STATEABBRV'].astype(str).str.lower().str.contains(q_state)]
            
            if not matches.empty:
                print(f"Found {len(matches)} matching result(s):")
                display(matches[search_cols].reset_index(drop=True))
            else:
                print(f"No records found matching County: '{q_county}' & State: '{q_state}'.")

    county_input.observe(update_search_results, names='value')
    state_input.observe(update_search_results, names='value')
    
    ui = widgets.VBox([
        widgets.HTML("<b>Enter County Name and State Abbreviation to Search:</b>"),
        county_input,
        state_input,
        search_output
    ])
    
    display(ui)

# 立即在 VS Code 中执行整个看板程序
run_fema_risk_dashboard()
```
